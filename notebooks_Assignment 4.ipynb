```python
with open("numbers.txt", "w") as file:
    for i in range(1, 6):
        file.write(f"{i}\n")
        
```

```python
total_sum = 0
with open("numbers.txt", "r") as file:
    for line in file:
        total_sum += int(line.strip())

print("Total Sum:", total_sum)
```

```python
filename = "numbers.txt"  # You can replace this with any file name
with open(filename, "r") as file:
    line_count = len(file.readlines())

print(f"Total lines in '{filename}':", line_count)
```

```python
# Q solution code here
# Explanation: The benefit of using the 'with' statement is that it automatically 
# closes the file after the block of code executes, even if an exception or error occurs. 
# This prevents memory leaks and file corruption.

try:
    with open("data.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("data.txt does not exist yet.")
    
```

```python
try:
    with open("sample.txt", "r") as file:
        lines_list = file.readlines()
    print(lines_list)
except FileNotFoundError:
    print("sample.txt does not exist.")
    
```

```python
try:
    with open("source.txt", "r") as source:
        data = source.read()
    
    with open("destination.txt", "w") as dest:
        dest.write(data)
    print("File copied successfully.")
except FileNotFoundError:
    print("source.txt not found.")
    
```

```python
try:
    val = int(input("Enter a denominator: "))
    res = 10 / val
    print("Result:", res)
except (ZeroDivisionError, ValueError) as e:
    print(f"An error occurred: {type(e).__name__}")
    
```

```python
try:
    number = int(input("Enter a valid integer: "))
except ValueError:
    print("That is not a valid integer.")
else:
    print(f"Success! You safely entered the number {number}.")
    
```

```python
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Access granted."

try:
    user_age = int(input("Enter your age: "))
    print(validate_age(user_age))
except ValueError as error:
    print("Validation Error:", error)
    
```

```python
file_handler = None
try:
    file_handler = open("sample_data.txt", "r")
    print(file_handler.read())
except FileNotFoundError:
    print("Error: 'sample_data.txt' does not exist.")
finally:
    if file_handler:
        file_handler.close()
        print("File stream closed safely.")
        
```
